```python
# from pandas dataframe
df = pd.read_csv('./dataset/ronaldo_vs_messi.csv')
df
```

```python
# now plotting graph
plt.plot(df['Year'],df['Ronaldo Goals'])
```

```python
# plotting multiple plots
plt.plot(df['Year'],df['Ronaldo Goals'])
plt.plot(df['Year'],df['Messi Goals'])
```

```python
# labels title
plt.plot(df['Year'],df['Ronaldo Goals'])
plt.plot(df['Year'],df['Messi Goals'])
plt.title('Ronaldo vs Messi (Goals)')
plt.xlabel('Year')
plt.ylabel('Goals')
```

```python
# colors , line and markers
plt.plot(df['Year'],df['Ronaldo Goals'],color = 'red',linestyle = 'dashed') #we can also provide hex code for color
plt.plot(df['Year'],df['Messi Goals'], color = 'blue',linestyle = 'dotted') #linestyle = dashed,dotted,solid(default),dashdot
plt.title('Ronaldo vs Messi (Goals)')
plt.xlabel('Year')
plt.ylabel('Goals')
```

```python
# marker
plt.plot(df['Year'],df['Ronaldo Goals'],color = 'red',marker = 'o' ,markersize = 5) #we can also provide hex code for color
plt.plot(df['Year'],df['Messi Goals'], color = 'blue',marker = 'o') #linestyle = dashed,dotted,solid(default),dashdot
plt.title('Ronaldo vs Messi (Goals)')
plt.xlabel('Year')
plt.ylabel('Goals')
```

```python
# using label and legend 
# till now in the graph we cannot say which one is the graph for ronaldo, and which one is for messi
plt.plot(df['Year'],df['Ronaldo Goals'],color = 'red',marker = 'o' ,markersize = 5,label = 'ronaldo') #we can also provide hex code for color
plt.plot(df['Year'],df['Messi Goals'], color = 'blue',marker = 'o',label = 'messi') #linestyle = dashed,dotted,solid(default),dashdot
plt.title('Ronaldo vs Messi (Goals)')
plt.xlabel('Year')
plt.ylabel('Goals')
plt.legend() #automatically in the place where there is empty space , we can modify this as well 
            # using loc = 'right','left','upper right','best' etc.
```

```python
# limiting axes
years = list(range(2005, 2026))
goals = [15, 25, 34, 35, 30, 48, 60, 63, 69, 61, 57, 55, 53, 49, 39, 44, 47, 16, 54, 43,1000]
# in this example lets assume cristiano scored 1000 goals in 2025 and now plot the graph
plt.plot(years,goals)
```

```python
# due to one outlier i.e 1000 graph appears like this and we can not visualize data properly
# cannot capture variation in data prperly due to high value of some(here one outlier 1000)
# we can limit this value as
plt.plot(years,goals)
plt.ylim(0,100) #goals only upto 100
```

```python
# we can limit on x axis also as
plt.plot(years,goals,marker = 'o') #to see goal points clearly (marker = 'o')
plt.ylim(0,100) #goals only upto 100
plt.xlim(2010,2018) #between peak of ronaldo
```

```python
# plt.grid() to show grid in the graph
plt.plot(df['Year'],df['Ronaldo Goals'],color = 'red',marker = 'o' ,markersize = 5,label = 'ronaldo')
plt.plot(df['Year'],df['Messi Goals'], color = 'blue',marker = 'o',label = 'messi') 
plt.title('Ronaldo vs Messi (Goals)')
plt.xlabel('Year')
plt.ylabel('Goals')
plt.legend()
plt.grid()
```

### Scatter Plots

- Bivariate analysis
- numerical vs numerical (can be done for categorical vs numerical but doesnot make sense that much)
- use case : Finding correlation (relation between two numerical values)
- example : house price prediction or any regression problem (best suited)

Scatter plot is just like a 2d line plot, 2d plot is a special case of scatter plot , we connect point by line in 2d line plot thats the difference

```python
# plotting simple scatter function
x = np.linspace(-10,10,50)
y = 7 * x + 5 + np.random.randint(1,200,50)
y
```

```python
plt.scatter(x,y)  #there is no kind parameter in matplotlib like in pandas dataframe
```

```python
plt.scatter(x,y,color = 'red',marker = 'x')
```

```python
# lets import ipl data
batter = pd.read_csv('./dataset/batter.csv')
batter
```

```python
# taking top 50 batsmen
batsmen = batter.head(50)
batsmen
```

```python
# plotting avg vs strike rate
plt.scatter(batsmen['avg'],batsmen['strike_rate'],color = 'orange')
plt.title('IPL top 50 batsmen`s avg vs strike rate')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
# size
plt.scatter(batsmen['avg'],batsmen['strike_rate'],color = 'orange',s = batsmen['avg']*7) # using s we can control size , here marker size is based on average
plt.title('IPL top 50 batsmen`s avg vs strike rate')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
# scatter plot using plot function
plt.plot(batsmen['avg'],batsmen['strike_rate'],'o')
```

- Using plot function -- faster
- using scatter function === slower

### Bar Chart

- Bivariate analysis
- Numerical vs categorical
- use case - Aggregate analysis of group

```python
# simple barchart
players = ['Ronaldo','Messi','Neymar','Benzema','Lewandowski']
ucl_goals = [140,129,43,90,103]
plt.bar(players,ucl_goals)
```

```python
# different color for each players goal
players = ['Ronaldo', 'Messi', 'Neymar', 'Benzema', 'Lewandowski']
ucl_goals = [140, 129, 43, 90, 103]
colors = ['red', 'blue', 'green', 'orange', 'purple']

plt.bar(players, ucl_goals, color=colors)

plt.xlabel('Players')
plt.ylabel('UCL Goals')
plt.title('UCL Goals by Player')
```

```python
# horizontal bar chart
players = ['Ronaldo', 'Messi', 'Neymar', 'Benzema', 'Lewandowski']
ucl_goals = [140, 129, 43, 90, 103]
colors = ['red', 'blue', 'green', 'orange', 'purple']

plt.barh(players, ucl_goals, color=colors)

plt.ylabel('Players')
plt.xlabel('UCL Goals')
plt.title('UCL Goals by Player')
```

```python
# multiple bar charts (important & very useful)
# let me load dataset
df = pd.read_csv('./dataset/player_goals_2015_2018.csv')
df
```

```python
# here if we want to draw bar chart of player goals in 2015 then we can do like:
plt.bar(df['Player'],df['2015'])
```

```python
# but what if we want the 2016 data also side by side ?
# we cannot plot multiple like in scatter plot or 2d line plot
# lets try what happens we do that
plt.bar(df['Player'],df['2015'])
plt.bar(df['Player'],df['2016'])  #only cristiano ronaldo has more goals in 2016 than in 2015 (bar are overlapping)
```

```python
# correct way of plotting multiple bar charts
plt.bar(np.arange(df.shape[0])-0.4,df['2015'],width = 0.2,color = 'red',label = '2015')  #created a array from 0 to no .of rows (df.shape[0]) for numerical cal
plt.bar(np.arange(df.shape[0])-0.2,df['2016'],width = 0.2,color = 'green',label = '2016')
plt.bar(np.arange(df.shape[0]),df['2017'],width = 0.2,color = 'blue',label = '2017')
plt.bar(np.arange(df.shape[0])+0.2,df['2018'],width = 0.2,color = 'yellow',label = '2018')
plt.xticks(np.arange(df.shape[0]),df['Player'])   #replace current x labels with new 
plt.legend()
plt.show()
```

```python
# using pandas it is very easier although pandas uses matplotlib internally
df.set_index('Player')
df.plot(kind = 'bar')
```

```python
# in the above matplotlib graph player name is not clearly visible as it is big :
# to fix this we can rotate value with certain angle as:
plt.bar(np.arange(df.shape[0])-0.4,df['2015'],width = 0.2,color = 'red',label = '2015')  #created a array from 0 to no .of rows (df.shape[0]) for numerical cal
plt.bar(np.arange(df.shape[0])-0.2,df['2016'],width = 0.2,color = 'green',label = '2016')
plt.bar(np.arange(df.shape[0]),df['2017'],width = 0.2,color = 'blue',label = '2017')
plt.bar(np.arange(df.shape[0])+0.2,df['2018'],width = 0.2,color = 'yellow',label = '2018')
plt.xticks(np.arange(df.shape[0]),df['Player'],rotation = -50)   #replace current x labels with new and rotation = angle 
plt.legend()
plt.show()
```

```python
# stacked bar chart
plt.bar(df['Player'],df['2015'],label = '2015')
plt.bar(df['Player'],df['2016'],bottom=df['2015'],label = '2016')
plt.bar(df['Player'],df['2017'],bottom=df['2015']+df['2016'], label = '2017')
plt.bar(df['Player'],df['2018'],bottom=df['2015']+df['2016']+df['2017'],label = '2018')
plt.legend()
plt.xticks(rotation = -50)
plt.show()
```

### Pie Chart

- Univariate/Bivariate analysis
- numerical vs categorical
- use case : to find contribution on a standard scale

```python
# lets load dataset
gayle = pd.read_csv('./dataset/gayle-175.csv') #ipl match data which contains all rcb batsmen run
gayle
```

```python
plt.pie(gayle['batsman_runs'],labels = gayle['batsman'])
plt.show()
```

```python
# seeing percentage
plt.pie(gayle['batsman_runs'],labels = gayle['batsman'],autopct='%0.1f%%')
plt.show()
```

```python
# explode (lets treat kohli differently)
plt.pie(gayle['batsman_runs'],labels = gayle['batsman'],autopct='%0.1f%%',explode = [0,0,0,0,0,0.1])
plt.show()
```

```python
# lets try with de villiers and kohli
plt.pie(gayle['batsman_runs'],labels = gayle['batsman'],autopct='%0.1f%%',explode = [0.1,0,0,0,0,0.1])
plt.show()
```
