```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## color scattered graphs

```python
#loading iris dataset
iris = pd.read_csv('./dataset/iris.csv')
iris.head()
```

```python
# lets plot it normally
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
```

```python
# lets make it more attrictive
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c = iris['Species'], cmap='winter')
```

```python
iris.sample(7)
```

```python
# first we need to convert target variable i.e species to numerical
iris['Species'] = iris['Species'].replace({'Iris-setosa' : 0, 'Iris-versicolor' : 1, 'Iris-virginica' : 2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c = iris['Species'], cmap='winter')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# alpha ----> opacity
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c = iris['Species'], cmap='winter',alpha = 0.3)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
#colorbar
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c = iris['Species'], cmap='jet',alpha = 0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

### Plot Size

```python
plt.figure(figsize=(10,6))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c = iris['Species'], cmap='jet',alpha = 0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

### Annotations

```python
#lets load dataset of ipl
ipl = pd.read_csv('./dataset/batter.csv')
ipl.head()
```

```python
# lets take top 50
batsmen = ipl.head(50)
batsmen.sample(5)
```

```python
# now plotting scatter plot
plt.scatter(batsmen['avg'],batsmen['strike_rate'],color = 'red')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.title('IPL TOP 50 Batsmen')
```

```python
#till now it is good but we cannot know the player from graph.
plt.figure(figsize=(20,15))
plt.scatter(batsmen['avg'],batsmen['strike_rate'],color = 'red')#s = batsmen['runs'])
plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.title('IPL TOP 50 Batsmen')

#looping for text annotation
for i in range(batsmen.shape[0]):
    plt.text(batsmen['avg'].values[i],batsmen['strike_rate'].values[i],batsmen['batter'].values[i])
plt.savefig('ipl.png')
```

### Horizontal and vertical line

this useful to divide graph vertically or horizontally

```python
# lets we have to locate a player with strike rate higher than 130
plt.figure(figsize=(20,15))
plt.scatter(batsmen['avg'],batsmen['strike_rate'],color = 'red')#s = batsmen['runs'])
plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.title('IPL TOP 50 Batsmen')

plt.axhline(130,color = 'green')
#looping for text annotation
for i in range(batsmen.shape[0]):
    plt.text(batsmen['avg'].values[i],batsmen['strike_rate'].values[i],batsmen['batter'].values[i])
plt.savefig('ipl.png')
```

```python
# we also want average greater than 35
plt.figure(figsize=(20,15))
plt.scatter(batsmen['avg'],batsmen['strike_rate'],color = 'red')#s = batsmen['runs'])
plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.title('IPL TOP 50 Batsmen')

plt.axhline(130,color = 'green')
plt.axvline(35,color = 'green')
#looping for text annotation
for i in range(batsmen.shape[0]):
    plt.text(batsmen['avg'].values[i],batsmen['strike_rate'].values[i],batsmen['batter'].values[i])
plt.savefig('ipl.png')
```

we can plot many lines

## Subplots

- till now we have plot graph only using using plt.func(),
- there is also another method using subplots

```python
fig,ax = plt.subplots()
ax.scatter(ipl['avg'],ipl['strike_rate'])
```

```python
plt.subplots()
```

- using plt.subplots() we get two object one is figure object and another is axes object we store that in two variable,
- then in that axes we can plot any kind of graphs

```python
fig,ax = plt.subplots(figsize = (10,6))
ax.scatter(ipl['avg'],ipl['strike_rate'])
ax.set_xlabel('Average') #xlabel ---> set_xlabel
ax.set_ylabel('Strike rate')
ax.set_title('Top 50 IPL Batsmen')
fig.show()   #plt.show() to fig.show()
```

```python
# multiple plots in single figure
fig,ax = plt.subplots(nrows=2,ncols=1)         #grid
ax  #array
```

```python
fig,ax = plt.subplots(nrows=2,ncols=1,sharex=True)
ax[0].scatter(ipl['avg'],ipl['strike_rate'])
ax[1].scatter(ipl['avg'],ipl['runs'])

ax[0].set_title('Average vs Strike Rate')
ax[0].set_ylabel('Strike Rate')

ax[1].set_title('Average vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Average')
fig.suptitle('Ipl Graphs',ha = 'center')   #title for figure 
```

```python
# 4 graphs in one figure
fig,ax = plt.subplots(nrows=2,ncols=2,figsize = (10,10))
ax[0,0].scatter(ipl['avg'],ipl['strike_rate'])
ax[0,1].scatter(ipl['avg'],ipl['runs'])
ax[1,0].hist(ipl['avg'])
ax[1,1].hist(ipl['runs'])

ax[0,0].set_title('Average vs Strike Rate')
ax[0,0].set_ylabel('Strike Rate')
ax[0,0].set_xlabel('Average')

ax[0,1].set_title('Average vs Runs')
ax[0,1].set_ylabel('Runs')
ax[0,1].set_xlabel('Average')

ax[1,0].set_title('Average histogram')
ax[1,1].set_title('runs histogram')
fig.suptitle('Ipl Graphs',ha = 'center')
```

```python
# third method
# using  plt.figure()
fig = plt.figure(figsize = (10,6))
ax1 = fig.add_subplot(2,2,1)
ax1.scatter(ipl['avg'],ipl['strike_rate'])

ax2 = fig.add_subplot(2,2,2)
ax2.scatter(ipl['avg'],ipl['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(ipl['avg'])

ax4 = fig.add_subplot(2,2,4)
ax4.hist(ipl['runs'])
```

### 3D line plot

```python
x = [0,10,30,25,25]
y = [0,20,50,70,40]
z = [0,30,70,65,0]
fig = plt.figure()
ax = plt.subplot(projection = '3d')
ax.scatter3D(x,y,z,color = 'green',s = [100,100,100,100,100])
ax.set_xlabel('x-axis')
ax.set_ylabel('y-axis')
ax.plot3D(x,y,z,color = 'red')
```

### 3d surface plot

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)
xx,yy
```

```python
z = xx**2 + yy**2
fig = plt.figure()
ax = plt.subplot(projection = '3d')
a = ax.plot_surface(xx,yy,z,cmap = 'viridis')
plt.colorbar(a)
```

### Contour plot

- Represent 3d graph in 2d

```python
fig = plt.figure()
ax = plt.subplot()
a = ax.contour(xx,yy,z)
plt.colorbar(a)
```

```python
# using contourf
fig = plt.figure()
ax = plt.subplot()
a = ax.contourf(xx,yy,z)
plt.colorbar(a)
```

above surface plot(3d) and countour plot represent a same plot

### Heatmap

- to plot grid to graph

```python
# lets see one ipl data
ball = pd.read_csv('./dataset/IPL_Ball_by_Ball_2008_2022.csv')
ball.head(5)
```

```python
ball.tail(5)
```

```python
delivery_data = ball[(ball['ballnumber'].isin([1,2,3,4,5,6])) & (ball['batsman_run'] == 6)]
delivery_data.head(5)
```

```python
delivery_data.tail()
```

```python
grid = delivery_data.pivot_table(index = 'overs',columns='ballnumber',values = 'batsman_run',aggfunc='count')
grid
```

```python
plt.figure(figsize=(10,6))
plt.imshow(grid)
plt.colorbar()
```
